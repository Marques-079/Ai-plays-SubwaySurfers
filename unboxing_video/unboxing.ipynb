```python
# Minimal Jupyter cell: checks if pixel (380, 900) in image "box_open"
# (same directory) is ~yellow RGBA(255,247,136,255) within ±25% tolerance per
# channel. Prints/returns a boolean.
#
# It searches for: ./box_open, ./box_open.png/.jpg/.jpeg/.bmp/.webp (and the
# same under /mnt/data) and uses the first one it finds.

import os, cv2, numpy as np

# --- config ---
X, Y = 380, 900
TOL = int(round(0.14 * 255))  # 25% per-channel tolerance => ~64
TARGET_BGRA = np.array([136, 247, 255, 255], dtype=np.int16)  # RGBA(255,247,136,255) -> BGRA


img_path = "/Users/marcus/Documents/Screenshot 2025-10-09 at 1.44.20 AM.png"
img_path =  "/Users/marcus/Documents/GitHub/Ai-plays-SubwaySurfers/replays/replay_000097.jpg"

def _check_pixel_yellow(img, x, y, tol=TOL):
    if img is None or img.size == 0:
        return False
    h, w = img.shape[:2]
    if not (0 <= x < w and 0 <= y < h):
        print(f"[warn] Pixel ({x},{y}) is outside image bounds {w}x{h}.")
        return False
    # Ensure 3 or 4 channels
    if img.ndim == 2:
        # grayscale; can't match RGBA robustly
        return False
    if img.shape[2] == 3:
        # BGR only
        b, g, r = img[y, x].astype(np.int16)
        diff = np.abs(np.array([b, g, r], dtype=np.int16) - TARGET_BGRA[:3])
        return bool(np.all(diff <= tol))
    elif img.shape[2] >= 4:
        # BGRA (or more; take first 4)
        b, g, r, a = img[y, x, :4].astype(np.int16)
        diff = np.abs(np.array([b, g, r, a], dtype=np.int16) - TARGET_BGRA)
        return bool(np.all(diff <= tol))
    else:
        return False

if img_path is None:
    print("[error] Could not find 'box_open' image in current dir or /mnt/data "
          "(tried no extension and common image extensions).")
    yellow_pixel_pass = None
else:
    img = cv2.imread(img_path, cv2.IMREAD_UNCHANGED)  # preserves alpha if present
    passed = _check_pixel_yellow(img, X, Y, TOL)
    yellow_pixel_pass = passed
    print(f"[check] path={img_path} | size={img.shape[1]}x{img.shape[0]} | "
          f"pixel({X},{Y})≈RGBA(255,247,136,255)±{TOL}? -> {passed}")

# Expose a simple boolean variable as requested:
yellow_pixel_pass
```
